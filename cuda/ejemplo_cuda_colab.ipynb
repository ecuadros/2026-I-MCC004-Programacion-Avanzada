```python
%%writefile ejemplo_cuda.cu

#include <cstdio>
#include <cuda_runtime.h>

// Esta funcion se ejecuta en la GPU
__global__ void sumar()
{
    int a = 5;
    int b = 3;

    printf("Suma realizada en la GPU: %d\n", a + b);
}

int main()
{
    // Ejecutar el kernel con un bloque y un hilo
    sumar<<<1, 1>>>();

    // Esperar a que la GPU termine
    cudaDeviceSynchronize();

    return 0;
}
```

```python
!nvcc ejemplo_cuda.cu -o ejemplo_cuda
```
